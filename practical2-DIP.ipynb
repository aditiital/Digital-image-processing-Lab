```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Read image
img = cv2.imread('/content/moon.webp', 0)
# Read image
img1 = cv2.imread('/content/MOON1.jpg', 0)
# Negative Transformation
negative = 255 - img
# Gamma Transformation
gamma = 0.5
gamma_image = np.array(255 * (img / 255) ** gamma, dtype='uint8')
# Display Images
plt.figure(figsize=(10,6))
plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original GrayScale")
plt.subplot(2,2,2)
plt.imshow(negative, cmap='gray')

plt.title("Negative")
plt.subplot(2,2,4)
plt.imshow(gamma_image, cmap='gray')
plt.title("Gamma Transform")
# Negative Transformation
negative1 = 255 - img1
# Gamma Transformation
gamma1 = 0.5
gamma_image1 = np.array(255 * (img1 / 255) ** gamma1, dtype='uint8')
# Display Images
plt.figure(figsize=(10,6))
plt.subplot(2,2,1)
plt.imshow(img1, cmap='gray')
plt.title("Original GrayScale")
plt.subplot(2,2,2)
plt.imshow(negative1, cmap='gray')
plt.title("Negative")
plt.subplot(2,2,4)
plt.imshow(gamma_image1, cmap='gray')
plt.title("Gamma Transform")
```
