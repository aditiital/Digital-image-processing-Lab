```python
import cv2
import os
import matplotlib.pyplot as plt
# Read Image
img = cv2.imread('/content/monster.jpg')
# Original File Size
original_size = os.path.getsize('/content/monster.jpg')
# Save Compressed Image
cv2.imwrite(
'compressed1.jpg',
img,
[cv2.IMWRITE_JPEG_QUALITY, 40]
)
# Compressed File Size
compressed_size = os.path.getsize('compressed1.jpg')

print("Original Size :", original_size, "bytes")
print("Compressed Size :", compressed_size, "bytes")
compression_ratio = original_size / compressed_size
print("Compression Ratio :", round(compression_ratio, 2))
# Read images for display
original_img = cv2.cvtColor(cv2.imread('monster.jpg'), cv2.COLOR_BGR2RGB)
compressed_img = cv2.cvtColor(cv2.imread('compressed1.jpg'), cv2.COLOR_BGR2RGB)
# Show comparison
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(original_img)
plt.title(f"Original\n{original_size:,} bytes")
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(compressed_img)
plt.title(f"Compressed\n{compressed_size:,} bytes")
plt.axis('off')
plt.tight_layout()
plt.show()
#aditi-ital(CD23012)
```
