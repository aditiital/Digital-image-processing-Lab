```python
import cv2
import matplotlib.pyplot as plt
# Read Image
img = cv2.imread('/content/swan.jpg')
# Convert BGR to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# RGB to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
# RGB to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Display Images
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(rgb)
plt.title("RGB Image")
plt.subplot(1,3,2)
plt.imshow(hsv)
plt.title("HSV Image")
plt.subplot(1,3,3)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
```
