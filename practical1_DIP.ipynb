```python
import cv2
import matplotlib.pyplot as plt
# Read image
img = cv2.imread('/content/red-moon.jpg', 0)
# Sampling (Resize)
sampled_img = cv2.resize(img, (200, 200))
# Quantization
levels = 16
quantized_img = (img // (256 // levels)) * (256 // levels)
# Display
plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.subplot(1,3,2)
plt.imshow(sampled_img, cmap='gray')
plt.title("Sampled")
plt.subplot(1,3,3)
plt.imshow(quantized_img, cmap='gray')
plt.title("Quantized")
```
