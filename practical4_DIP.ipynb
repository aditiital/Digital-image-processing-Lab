```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('/content/panda.webp', 0)

print(img.shape)

# Gaussian smoothing
smooth = cv2.GaussianBlur(img, (5, 5), 0)

# Low-pass blur
blur = cv2.GaussianBlur(img, (11, 11), 0)

# High-pass component
high_pass = cv2.subtract(img, blur)

# Sharpened image
sharpened = cv2.add(img, high_pass)

# Display
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(smooth, cmap='gray')
plt.title("Smoothed")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(sharpened, cmap='gray')
plt.title("Sharpened")
plt.axis('off')

plt.show()
```
